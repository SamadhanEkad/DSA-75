```python
n = 15
result = []

for i in range(1,n+1):
    if n % i == 0:
        result.append(i)

print(result)
```

```python
# Using Function
```

```python
def factors1(n):
    result = []

    for i in range(1,n+1):
        if n % i == 0:
            result.append(i)

    return(result)
```

```python
print(factors1(15))
```

```python
# Better Solution
```

```python
def factors2(n):
    result = []

    for i in range(1,n//2):
        if n % i == 0:
            result.append(i)

    result.append(n)
    return(result)
```

```python
print(factors2(n))
```

```python
# Optimal Solution
```

```python
import numpy as np

def factors3(n):
    result = []

    for i in range(1,int(np.sqrt(n))+1):
        if n % i == 0:
            result.append(i)
        if n // i != i:
            result.append(n//i)

    result.sort()
    return result
```

```python
print(factors3(15))
```
