### Method : 01

```python
nums = [5,6,7,7,9,111,1,5,1,1]
dict = {}                       # Empty Dictionary / Freq map

for i in range(0,len(nums)):
    if nums[i] in dict:
        dict[nums[i]] += 1
    else:
        dict[nums[i]] = 1

print(dict[nums[0]])
```

### Method : 03

```python
nums = [5,6,7,7,9,111,1,5,1,1]
hash_map = {}
n = len(nums)

for i in range(0,n):
    hash_map[nums[i]] = hash_map.get(nums[i],0) + 1
```
