### 1.Two Sum
**Input:**
```text
nums = [2, 7, 11, 15]
target = 9

output : [0,1]

````python
# Brute Force Solution
def two_sum(arr,target):
    n = len(arr)
    for i in range(0,n-1):
        for j in range(i+1,n):
            if arr[i] + arr[j] == target:
                return [i,j]
````

````python
two_sum([2,7,11,15],9)
````

````python
# Optimal solution
def two_sum(arr,target):
    n = len(arr)
    hash_map = {}
    for i in range(0,n):
        remaining = target - arr[i]
        if remaining in hash_map:
            return [hash_map[remaining],i]
        hash_map[arr[i]] = i
````

````python
two_sum([5,9,1,2,4,15,6,3],13)
````
