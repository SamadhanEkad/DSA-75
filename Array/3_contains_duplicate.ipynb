### Method: 01

```python
def duplicate(arr):
    n = len(arr)

    for i in range(n):
        for j in range(i+1,n):
            if arr[i] == arr[j]:
                return True
    
    return False
```

```python
duplicate([1,2,3,4])
```

### Method : 02

```python
# Optimal Solution 
def duplicate(arr):
    return len(arr) != len(set(arr))
```

```python
duplicate([1,2,3,4])
```

```python
duplicate([1,2,3,4,1])
```
