### Contains Duplicate II

**Difficulty:** Easy

**Problem:** Given an integer array `nums` and an integer `k`, return `true` if two distinct indices `i` and `j` satisfy `nums[i] == nums[j]` and `abs(i - j) <= k`. Otherwise, return `false`.

**Example:**
```python
Input: nums = [1, 2, 3, 1], k = 3
Output: true
```

````python
def duplicate(nums,k):
    n = len(nums)
    d = {}
    for i in range(n):
        if nums[i] in d and abs(i-d[nums[i]]) <= k:
            return True
        else:
            d[nums[i]] = i
````

````python
duplicate([1,2,3,1],3)
````
