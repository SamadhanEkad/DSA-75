### 1. Store Frequency in dictionary

#### Method : 01


```python
nums = [5,6,7,7,9,111,1,5,1,1]
dict = {}                       # Empty Dictionary / Freq map

for i in range(0,len(nums)):
    if nums[i] in dict:
        dict[nums[i]] += 1
    else:
        dict[nums[i]] = 1

print(dict[nums[0]])
```

#### Method:02

```python
nums = [5,6,7,7,9,111,1,5,1,1]
hash_map = {}
n = len(nums)

for i in range(0,n):
    hash_map[nums[i]] = hash_map.get(nums[i],0) + 1

print(hash_map)
```

### Introduction of Hashing:

```python
n = [5,3,2,2,1,5,5,7,5,10]
m = [10,111,1,9,5,67,2]

for num1 in m:
    count = 0

    for num2 in n:
        if num2 == num1:
            count += 1

    print(num1, ":", count)
```

```python
n = [5,3,2,2,1,5,5,7,5,10]
m = [10,111,1,9,5,67,2]

hash_list = [0] * 11

for num1 in n:
    hash_list[num1] += 1

for num2 in m:
    if num2<1 or num2>10:
        print(0)
    else:
        print(num2,":",hash_list[num2])
```

```python
# Using Dictionary Approch (Digit Hashing)
n = [5,3,2,2,1,5,5,7,5,10]
m = [10,111,1,9,5,67,2]

hash_map = {}
for num in n:
    hash_map[num] = hash_map.get(num,0) + 1

for num in m:
    print(num,":",hash_map.get(num,0))
```

```python
# Character Hashing
s = "aabgdtwuydnxgesqooobs"
q = ['a','d','y','x']

hash_list = [0] * 26

for ch in s:
    ascii_val = ord(ch)
    index = ascii_val - 97
    hash_list[index] += 1

for ch in q:
    ascii_val = ord(ch)
    index = ascii_val - 97
    print(ch,":",hash_list[index])
```
