### 1. Store Frequency using dictionary

```python
def frequency(arr):
    n = len(arr)
    freq = {}
    for i in range(0,n):
        freq[arr[i]] = freq.get(arr[i],0) + 1

    return freq
```

```python
frequency([1,2,2,3,4,5,4,3,3,4,5,2,1])
```

### 2. Reverse an Array

```python
def reverse(arr):
    n = len(arr)
    result = []
    for i in range(n-1,-1,-1):
        result.append(arr[i])

    return result
```

```python
reverse([1,2,3,4,5])
```
