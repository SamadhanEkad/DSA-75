### 1. Palindrome Number

```python
def palindrome(n):
    num = n
    rev = 0
    while num > 0:
        digit = num % 10
        rev = rev * 10 + digit
        num = num // 10

    return n == rev
```

```python
palindrome(12121)
```

### 2. Armstrong Number

```python
def armstrong(n):
    num = n
    l = int(len(str(n)))
    total = 0
    while num > 0:
        digit = num % 10
        total += digit ** l
        num = num // 10

    return n == total
```

```python
armstrong(153)
```

### 3. Print all factors of give number

```python
def factors(n):
    result = []
    for i in range(1,n+1):
        if n % i == 0:
            result.append(i)

    return result
```

```python
factors(28)
```

### 4. Factorial of Number

```python
def factorial(n):
    fact = 1
    for i in range(1,n):
        fact = fact * n
        n = n - 1

    return fact
```

```python
factorial(5)
```

### 5. Fibonacci series of number

```python
def fibonacci(n):
    x = 0
    y = 1
    z = 0
    for i in range(n):
        print(x , end= " ")
        z = x + y
        x = y
        y = z
```

```python
fibonacci(10)
```
