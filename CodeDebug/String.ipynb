### 1. Palindrome String

```python
def palindrome(s):
    return s == s[::-1]
```

```python
palindrome("madam")
```

```python
def palindrome(s):
    n = len(s)
    left = 0
    right = n-1
    while left < right:
        if s[left] != s[right]:
            return False

        left += 1
        right -= 1

    return True
```

```python
palindrome("madam")
```
