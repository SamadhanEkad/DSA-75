### 1. Two Sum II – Input Array Is Sorted

Given a **1-indexed sorted array** `numbers`, find two numbers whose sum equals `target`.

Return their **1-based indices** as `[index1, index2]`.

**Conditions:**
- Exactly one solution exists.
- Cannot use the same element twice.
- Use O(1) extra space.

**Example 1:**
- Input: `numbers = [2, 7, 11, 15], target = 9`
- Output: `[0, 1]`

```python
# Brute-Force Approch
def two_sum(arr,target):
    n = len(arr)
    
    for i in range(n-1):
        for j in range(i+1,n):
            total = arr[i] + arr[j]
            if total == target:
                return [i,j]
```

```python
two_sum([2,7,11,15],9)
```

```python
# Two pointer Approch
def two_sum(arr,target):
    l = 0
    r = len(arr)-1

    while l < r:
        total = arr[l] + arr[r]
    
        if total == target:
            return [l,r]
        elif total < target:
            l += 1
        else:
            r -= 1
    
    return []
```

```python
two_sum([2,7,11,15],13)
```

### 2. 3Sum (Medium)

Given an integer array `nums`, return all **unique triplets** `[nums[i], nums[j], nums[k]]` such that:

- `i != j`, `i != k`, and `j != k`.
- `nums[i] + nums[j] + nums[k] == 0`.
- No duplicate triplets are allowed.

**Example 1:**
Input: `nums = [-1,0,1,2,-1,-4]`  
Output: `[[-1,-1,2],[-1,0,1]]`

**Example 2:**
Input: `nums = [0,1,1]`  
Output: `[]`

**Example 3:**
Input: `nums = [0,0,0]`  
Output: `[[0,0,0]]`

```python
# Brute-Force Solution -> O(n^3)
def three_sum(arr):
    n = len(arr)
    my_set = set()
    for i in range(n):
        for j in range(i+1, n):
            for k in range(j+1, n):
                if arr[i] + arr[j] + arr[k] == 0:
                    temp = [arr[i], arr[j], arr[k]]
                    temp.sort()
                    my_set.add(tuple(temp))

    return [list(ans) for ans in my_set]
```

```python
three_sum([-1,0,1,2,-1,-4])
```

```python
# Optimum Solution -> O(NlogN) + O(N^2)
def three_sum(arr):
    ans = []
    n = len(arr)
    arr.sort()
    for i in range(n):
        if i!=0 and arr[i] == arr[i-1]:
            continue

        # Moving two pointers
        j = i+1
        k = n-1
        while j < k:
            total = arr[i] + arr[j] + arr[k]
            if total < 0:
                j += 1
            elif total > 0:
                k -= 1
            else:
                temp = [arr[i],arr[j],arr[k]]
                ans.append(temp)
                j += 1
                k -= 1
                # Skip the duplicate if occured
                while j < k and arr[j] == arr[j-1]:
                    j += 1
                while j < k and arr[k] == arr[k+1]:
                    k -= 1
    return ans
```

```python
three_sum([-1,0,1,2,-1,-4])
```
