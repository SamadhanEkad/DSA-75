### 1. Two Sum II – Input Array Is Sorted

Given a **1-indexed sorted array** `numbers`, find two numbers whose sum equals `target`.

Return their **1-based indices** as `[index1, index2]`.

**Conditions:**
- Exactly one solution exists.
- Cannot use the same element twice.
- Use O(1) extra space.

**Example 1:**
- Input: `numbers = [2, 7, 11, 15], target = 9`
- Output: `[1, 2]`

```python
# Brute-Force Approch
def two_sum(arr,target):
    n = len(arr)
    
    for i in range(n-1):
        for j in range(i+1,n):
            total = arr[i] + arr[j]
            if total == target:
                return [i+1,j+1]
```

```python
two_sum([2,7,11,15],9)
```

```python
# Two pointer Approch
def two_sum(arr,target):
    l = 0
    r = len(arr)-1

    while l < r:
        total = arr[l] + arr[r]
    
        if total == target:
            return [l+1,r+1]
        elif total < target:
            l += 1
        else:
            r -= 1
    
    return []
```

```python
two_sum([2,7,11,15],13)
```
